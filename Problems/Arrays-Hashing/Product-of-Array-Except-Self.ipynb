## Description

Given an integer array nums, return an array output where output[i] is the product of all the elements of nums except nums[i].

Each product is guaranteed to fit in a 32-bit integer.

Follow-up: Could you solve it in  O(n) time without using the division operation?

## My First Solution

First I solved it brute force and that was O(n^2) and really easy so there is no reason to put that here.

This solution starts with creating a result, left, and right array set to 1. These are the arrays I construct and make the solution.

- First we move through the array from left to right and at each index we create the sum of all the numbers to the left except the self.
- Next we move through the array from the right to the left, and we do the exact same thing. We create the product of all the items to the right except the self.
- From there we create the result array that we need to end up returning. We loop through the length of the original array and create a product of left[i] and right[i]. So at each index this creates the product of everything to the left and everything to the right without actually using self.
- This satisfies the O(n) requirement because it only ever goes through the length of the array
- With those two arrays it helps us construct the solution.



```python
class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:

        res = len(nums) * [1]
        left = len(nums) * [1]
        right = len(nums) * [1]

        for i in range(len(nums)):
            if i != 0:
                left[i] = nums[i-1] * left[i-1]

        for i in range(len(nums)-1, -1, -1):
            if i != len(nums)-1:
                right[i] = nums[i+1] * right[i+1]

        for i in range(len(res)):
            res[i] = left[i] * right[i]

        return res
```
