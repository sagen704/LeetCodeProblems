## Description

Design an algorithm to encode a list of strings to a string. The encoded string is then sent over the network and is decoded back to the original list of strings.

## My First Solution

Encode Section
- This is done by creating one encoded string with the length of the word and some sort of known delimiter
- So when you reconstruct the word all you need to do is to parse to get the length of the word until the delimiter and then you know how long to keep parsing to get the whole word

```python
class Solution:

    def encode(self, strs: List[str]) -> str:

        encoded_str = ""

        for word in strs:
            encoded_str = encoded_str + str(len(word)) + "#" + word

        return encoded_str

    def decode(self, s: str) -> List[str]:
        print(s)
        res = []
        i = 0
        while i < len(s):
            j = i

            while s[j] != "#":
                j += 1

            length_word = int(s[i:j])
            # move window
            j += 1
            res.append(s[j: j+length_word])

            i = j+length_word

        return res
```
