## Description

You are given an array of integers temperatures where `temperatures[i]` represents the daily temperatures on the ith day.

Return an array result where `result[i]` is the number of days after the `ith` day before a warmer temperature appears on a future day. If there is no day in the future where a warmer temperature will appear for the `ith` day, set `result[i]` to 0 instead.

## My First Solution

In this problem you were expected to use a stack, especially a monotonic decreasing stack. That is a stack where all the elements are in not strictly decreasing order and once you encounter a higher value in the array you pop until it is empty or it is less than the top value on the stack.

The tricky part is remembering you need to keep track of the indexes. Once you get the monotonic stack coded you need to update every index that you pop off.

- The result is initialized to all zero so that you can directly update indexes.
- Once you encounter a higher value than the top of the stack you need to start
  - Popping the value off the top and updating the `result[number you found higher index] = higher_number_index - number_index`
- After clearing it out you add on the new value

```python
class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:

        res = [0] * len(temperatures)

        stack = []

        for i in range(len(temperatures)):

            if not stack:
                stack.append([temperatures[i], i])

            # add it if it is decreasing or equal
            elif temperatures[i] <= stack[-1][0]:
                stack.append([temperatures[i], i])

            # remove if it is higher or equal
            else:
                while len(stack) != 0 and temperatures[i] > stack[-1][0]:
                    found_higher = stack.pop()
                    # print(f"found higher {found_higher}")

                    # updating the location
                    res[found_higher[1]] = i - found_higher[1]
                stack.append([temperatures[i], i])

        return res
```

## My Second Solution

I realized my first `if` and `elif` were practically useless and could be completely removed. The code still runs in the same big O time but it is much cleaner.

```python
class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:

        res = [0] * len(temperatures)

        stack = []

        for i in range(len(temperatures)):

            while len(stack) != 0 and temperatures[i] > stack[-1][0]:
                found_higher = stack.pop()
                # print(f"found higher {found_higher}")
                # updating the location
                res[found_higher[1]] = i - found_higher[1]
            stack.append([temperatures[i], i])

        return res
```
