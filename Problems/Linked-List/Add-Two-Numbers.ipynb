## Description

You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list.

You may assume the two numbers do not contain any leading zero, except the number 0 itself.

## My First Solution

This loops through and creates the number for the first linked list and then loops through for the second number. Adds the two numbers together and linked Listifies the answer

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:

        num1 = ""
        num2 = ""

        curr = l1
        while curr:
            num1 = str(curr.val) + num1
            curr = curr.next

        curr = l2
        while curr:
            num2 = str(curr.val) + num2
            curr = curr.next

        ans = str(int(num1) + int(num2))

        dummy = ListNode()
        curr = dummy

        for i in range(len(ans) - 1, -1, -1):
            curr.next = ListNode(int(ans[i]))
            curr = curr.next

        return dummy.next
```
