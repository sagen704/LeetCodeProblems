## Description

You are given a string s consisting of only uppercase english characters and an integer k. You can choose up to k characters of the string and replace them with any other uppercase English character.

After performing at most k replacements, return the length of the longest substring which contains only one distinct character.

## My First Solution

So I needed some help on this one and I needed to learn that you use a hash map for all the frequencies of the letters.

It follows the original formula of sliding window problems where I declare a left pointer and move through the data structure with a for loop. and shrink it when it does not meet a condition.

- Growing the window consisted with adding to the letter frequency dictionary/ updating the frequency.
- The formula followed was (length of window) - (largest frequency of letters) > k:
  - This means that I would shrink the window once I was not able to change all of the letters to one singular letter.
  -  So I wanted to check at every point of the window if I could change all of the letters to the same. The largest frequency of letters was the letter you would change the rest to so subtracting the window length from that and comparing with k let me see check if I could change all the letters to the same.
- Lastly I updated the answer by doing the sum of all the values in the window because that is the length of the window

```python
class Solution:
    def characterReplacement(self, s: str, k: int) -> int:

        character_frequencies = {}
        longest_substring = 0

        l = 0

        for r in range(len(s)):

            # Growing the window
            if s[r] not in character_frequencies:
                character_frequencies[s[r]] = 1
            else:
                character_frequencies[s[r]] = character_frequencies[s[r]] + 1

            # Shrinking the window
            while (r - l + 1) - max(character_frequencies.values()) > k:
                character_frequencies[s[l]] = character_frequencies[s[l]] - 1
                l += 1


            # Update the answer
            total = sum(character_frequencies.values())
            print(character_frequencies, longest_substring, total)

            if total > longest_substring:
                longest_substring = total

        return longest_substring
```
