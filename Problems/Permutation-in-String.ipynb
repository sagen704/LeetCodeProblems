## Description

Given two strings `s1` and `s2`, return true if `s2` contains a permutation of `s1`, or false otherwise.

In other words, return true if one of `s1`'s permutations is the substring of `s2`.


Example 1:

Input: `s1 = "ab"`, `s2 = "eidbaooo"`
Output: `true`
Explanation: s2 contains one permutation of s1 ("ba").
Example 2:

Input: `s1 = "ab"`, `s2 = "eidboaoo"`
Output: `false`


## My First Solution

For this solution I wanted to use a window the size of `s1` and check every window that size in `s2`. I first checked the edge case where len(s1) is greater than len(s2) where there will never be a permutation of that size.

Next I created dictionaries of `{'a': 0, 'b': 0 , ...}` for the counts of `s1` and `s2`. Then I filled the counts for those two dictionaries for counts of the first window of size `s1`.

Next I needed to get the number of matches for the first set of count windows. This solution uses matches so it is more optimal.

The next part is the sliding window part. Here I slide the window over one character every time and if I reach a permutation (matches are equal to 26) then I return true. Else I would update the right pointer.
- For the right pointer I would update the `s2` counts at the new right pointer. If I created a new match `s1_counts[k] == s2_counts[k]:` I would increment the matches by one. But if I moved the match to be one more `s1_counts[k] + 1 == s2_counts[k]`: then I destroyed a match and I needed to decrement the matches.
- Next I needed to update the left pointer. It is very similar to the right pointer where I decrement the left side because we are moving the window over. Then I check `s1_counts[lk] == s2_counts[lk]` to see if I made a new match. Or I might have destroyed a match if `s1_counts[lk] - 1 == s2_counts[lk]` it is now one less than what it was before. I need to rememeber to move left over at each window section

I lastly `return matches == 26` to check the last permutation window

```python
class Solution:
    def checkInclusion(self, s1: str, s2: str) -> bool:

        if len(s1) > len(s2):
            return False

        s1_counts = {chr(i + ord('a')) : 0 for i in range(26)}
        s2_counts = {chr(i + ord('a')) : 0 for i in range(26)}

        for i in range(len(s1)):
            s1_counts[s1[i]] += 1
            s2_counts[s2[i]] += 1

        matches = 0

        for i in range(26):
            letter = chr(i + ord('a'))
            if s1_counts[letter] == s2_counts[letter]:
                matches += 1

        # Sliding window portion

        l = 0

        for r in range(len(s1), len(s2)):

            if matches == 26: return True
            # update the count at right pointer
            k = s2[r]
            s2_counts[k] += 1
            if s1_counts[k] == s2_counts[k]:
                matches += 1
            elif s1_counts[k] + 1 == s2_counts[k]:
                matches -= 1

            # update the count at the left pointer
            lk = s2[l]
            s2_counts[lk] -= 1
            if s1_counts[lk] == s2_counts[lk]:
                matches += 1
            elif s1_counts[lk] - 1 == s2_counts[lk]:
                matches -= 1
            l += 1

        return matches == 26
```
