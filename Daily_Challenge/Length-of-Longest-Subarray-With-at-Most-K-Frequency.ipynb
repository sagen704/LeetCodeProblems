## Description

You are given an integer array `nums` and an integer `k`.

The frequency of an element `x` is the number of times it occurs in an array.

An array is called good if the frequency of each element in this array is less than or equal to `k`.

Return the length of the longest good subarray of `nums`.

A subarray is a contiguous non-empty sequence of elements within an array.

## My First Solution

So I needed to loop through the list on numbers and keep track of there frequencies. It is a basic sliding window that I update the window size and when the number at the right of the window contains more values then k then I need to shrink from the left. I only need to check at the r place in the array because that is where I recently updated the number of frequencies. After each index I update the result by seeing if the window is larger than the current result.

```python
class Solution:
    def maxSubarrayLength(self, nums: List[int], k: int) -> int:

        num_freq = {}
        res = 0
        l = 0

        for r in range(len(nums)):
            num_freq[nums[r]] = num_freq.get(nums[r], 0) + 1

            while num_freq[nums[r]] > k:
                num_freq[nums[l]] -= 1
                l += 1
            res = max(res, r-l+1)
        return res
```
