# Reorder Linked List

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
head = ListNode(2)
head.next = ListNode(4)
head.next.next = ListNode(6)
head.next.next.next = ListNode(8)
```

```python
fast = head.next
slow = head

# goes and makes slow the middle
while fast and fast.next:
    fast = fast.next.next
    slow = slow.next
```

```python
print(slow.val)
```

```python
tmp = slow.next
slow.next = None
prev = None

while tmp:
    print(prev)
    nxt = tmp.next
    tmp.next = prev
    prev = tmp
    tmp = nxt
```

```python
print(prev.val)
```
